```python
config = BacktestConfig(fast_window=20, slow_window=60, fee_bps=1.0, slippage_bps=1.0)
result = run_backtest(data, config)
metrics = performance_metrics(result)
metrics
```

## 미래정보 참조 점검

전략은 `signal.shift(1)`을 사용합니다. 따라서 오늘 종가로 계산한 신호가 오늘 수익률에 바로 적용되지 않고 다음 시점부터 포지션에 반영됩니다.


```python
result[['Close', 'fast_ma', 'slow_ma', 'signal', 'position', 'asset_return', 'strategy_return']].tail(10)
```

```python
result[['strategy_equity', 'buy_hold_equity']].plot(figsize=(10,5), title='추세추종 전략 vs Buy & Hold')
```
